# 01 — Data Acquisition & Cleaning

A thin narrative over `matchodds.data`: load the canonical master matches table (built by
`make data`) and sanity-check its coverage and distributions. All real logic lives in the
`matchodds` package — this notebook only reads and summarizes.

```python
from matchodds.data import matches

df = matches.load()
print(f"{len(df):,} matches | {df['date'].min()} -> {df['date'].max()}")
df.head()
```

## Coverage per league

```python
df["league"].value_counts()
```

## Coverage per season and league

```python
df = df.assign(season=df["date"].map(lambda d: d.year if d.month >= 7 else d.year - 1))
df.groupby(["season", "league"]).size().unstack(fill_value=0)
```

## Sanity checks

```python
print("Outcome distribution (H/D/A):", df["result"].value_counts(normalize=True).round(3).to_dict())
print(f"Missing closing-odds rate: {df['odds_home'].isna().mean():.1%}")
print(f"Mean goals - home {df['ft_home_goals'].mean():.2f}, away {df['ft_away_goals'].mean():.2f}")
df["ft_home_goals"].value_counts().sort_index()
```
